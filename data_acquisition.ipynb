# Data Acquisition

CS 418 football project. Loads all four of our datasets into dataframes.

**Research questions:**
- Is rushing or passing more effective in winning NFL games?
- Do the stats suggest punting or going for it on 4th down?
- How much does average WR separation correlate with receiving yardage?

All four datasets come from [nflverse](https://github.com/nflverse/nflreadpy),
accessed through the `nflreadpy` package (one shared import below instead of
repeating it per dataset).

```python
import polars as pl
import nflreadpy as nfl
```

## Primary dataset 1: Play-by-play (`playbyplay`)

One row = one play from a 2025 game. Notable columns: `posteam`, `yards_gained`,
`success`, `passer`, `rusher`, `receiver`.

```python
playbyplay = nfl.load_pbp(2025)
playbyplay.shape
```

```python
playbyplay.describe()
```

## Primary dataset 2: Player weekly stats (`player_stats`)

One row = one player's stat line for one week of one season (2021-2025).
Includes `rushing_*` and `passing_*` columns side by side -- the main dataset
for comparing rushing vs. passing production at the player level.

```python
player_stats = nfl.load_player_stats([2021, 2022, 2023, 2024, 2025])
player_stats.shape
```

```python
player_stats.head()
```

```python
# Columns relevant to rushing vs. passing
cols = [c for c in player_stats.columns if c.startswith('rushing_') or c.startswith('passing_')]
cols
```

```python
player_stats.select(cols).describe()
```

```python
player_stats.dtypes
```

## Secondary dataset 1: Team stats (`team_stats`)

One row = one team's cumulative stats for the 2025 regular season.
Notable columns: `passing_yards`, `rushing_yards`, `passing_tds`, `rushing_tds`.

```python
# Loads team-level stats for a given season/season-type ("reg" or "post")
team_stats = nfl.load_team_stats(2025, "reg")
team_stats.shape
```

```python
team_stats.describe()
```

## Secondary dataset 2: Next Gen Stats receiving (`wr_receiving`)

NFL Next Gen Stats receiving data, filtered down to 2025 regular-season wide
receiver weekly observations (the dataset also contains non-WR positions and
season-summary rows, so we narrow to what this project needs). Notable
columns: `avg_separation`, `avg_cushion`, `targets`, `receptions`, `yards`,
`avg_yac`. Will be used to investigate how average WR separation relates to
receiving yardage -- that analysis itself belongs in the EDA checkpoint, not here.

```python
ngs_receiving = nfl.load_nextgen_stats(
    seasons=2025,
    stat_type="receiving"
)

ngs_receiving.shape
```

```python
ngs_receiving.columns
```

```python
wr_receiving = (
    ngs_receiving
    .filter(
        (pl.col("season") == 2025) &
        (pl.col("season_type") == "REG") &
        (pl.col("player_position") == "WR") &
        (pl.col("week") > 0)
    )
    .select([
        "player_gsis_id",
        "player_display_name",
        "player_position",
        "team_abbr",
        "season",
        "week",
        "avg_cushion",
        "avg_separation",
        "avg_intended_air_yards",
        "targets",
        "receptions",
        "catch_percentage",
        "yards",
        "rec_touchdowns",
        "avg_yac",
        "avg_expected_yac",
        "avg_yac_above_expectation"
    ])
)

wr_receiving.shape
```
